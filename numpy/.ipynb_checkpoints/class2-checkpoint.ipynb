```python
# Normalization

import numpy as np

X=np.array([[2,80,1.75,30],
           [3,60,1.60,25],
           [1,95,1.80,40],
           [4,70,1.65,35],
           [2,88,1.72,28]])
X_min=X.min(axis=0)
X_max=X.max(axis=0)
print(X_min,X_max)

# normalise - broadcasting handles(5,4)/(4,) 
X_norm=(X-X_min)/(X_max-X_min)
print(X_norm.round(3))


# standardisation: mean=0, std=1 (Z-score|)
X_std = (X-X.mean(axis=0))/X.std(axis=0)
print('Mean after std : ', X_std.mean(axis=0))

```

```python
a=np.array([[4.0,7.0],
           [2.0,6.0]])

# Determinant
print(np.linalg.det(a))

# inverse
a_inv=np.linalg.inv(a)
print(a_inv)

print(a@a_inv) # for verify a@a_inv is idendtity

# a=np.eye((4),dtype=int)
# print(a)
```

```python
# eigenvalue and eigenvactor

a=np.array([[4.0,7.0],
           [2.0,6.0]])
eigenvalue,eigenvactors=np.linalg.eig(a)
print('eigenvalue : ',eigenvalue)
print('eigenvactor : ',eigenvactors)

# SVD (singular value Decompodition)
u,s,v=np.linalg.svd(a)
print('singlular value : ',s)

# Rank matrix
```

```python
# solve for x in: Ax=b
# 2x+y=5
# x+3y=10

a=np.array([[2,1],
           [1,3]])
b=np.array([5,10])

x=np.linalg.solve(a,b)
print(x)

# verify : a@x=b
print(a@x)
```

```python
# 2x+y+z=7
# x+3y+z=10
# x+y+2z=8

a=np.array([[2,1,1],
           [1,3,1],
           [1,1,3]])
b=np.array([7,10,8])
x=np.linalg.solve(a,b)
print(x)

# verify
print(a@x)
```

```python
a=np.array([[1,1,2],
           [2,1,3],
           [3,2,1]])
b=np.array([6,14,2])
x=np.linalg.solve(a,b)
print(x)

# verify
print(a@x)
```

```python
import numpy as np
 
# ── Single array 
arr = np.arange(100).reshape(10, 10)
 
# Save as binary .npy (fast, lossless)
np.save('my_array.npy', arr)
loaded = np.load('my_array.npy')
print(loaded.shape)
 
# Save as text (human readable)
np.savetxt('my_array.csv', arr, delimiter=',', fmt='%d')
loaded_txt = np.loadtxt('my_array.csv', delimiter=',')
 
# ── Multiple arrays ───────────────────────────────────
X = np.random.rand(100, 5)
y = np.random.randint(0, 3, 100)
 
# Save multiple arrays in a single .npz archive
np.savez('dataset.npz', features=X, labels=y)
 
# Load the archive
data = np.load('dataset.npz')
print(data['features'].shape)  # (100, 5)
print(data['labels'].shape)    # (100,)
 
# Compressed version (smaller file, slower read)
np.savez_compressed('dataset_compressed.npz', features=X, labels=y)
```
