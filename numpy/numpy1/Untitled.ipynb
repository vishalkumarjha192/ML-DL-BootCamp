```python
# Common imports used across problems
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
from sklearn.datasets import make_regression
X, y = make_regression(n_samples=200, n_features=1,
                       noise=15.0, random_state=42)
# X has shape (200, 1); y has shape (200,)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression().fit(X, y)
print('slope:', model.coef_[0], 'intercept:', model.intercept_)
# TODO: plt.scatter(...) then plt.plot(X, model.predict(X))
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Create model
model = LinearRegression()

# Fit model
model.fit(X, y)

# Print slope & intercept
print("Slope (coef):", model.coef_[0])
print("Intercept:", model.intercept_)

# Predictions
y_pred = model.predict(X)

# Plot
plt.scatter(X, y, label="Data Points")
plt.plot(X, y_pred, color='red', label="Fitted Line")
plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear Regression Fit")
plt.legend()
plt.show()
```

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import numpy as np

r2 = r2_score(y, y_pred)
rmse = np.sqrt(mean_squared_error(y, y_pred))
mae = mean_absolute_error(y, y_pred)

print("R²:", r2)
print("RMSE:", rmse)
print("MAE:", mae)
```

```python
# Residuals
residuals = y - y_pred

# RMSE (manual)
rmse_manual = np.sqrt(np.mean(residuals**2))

# R² (manual)
ss_res = np.sum(residuals**2)
ss_tot = np.sum((y - np.mean(y))**2)

r2_manual = 1 - (ss_res / ss_tot)

print("Manual RMSE:", rmse_manual)
print("Manual R²:", r2_manual)
```
