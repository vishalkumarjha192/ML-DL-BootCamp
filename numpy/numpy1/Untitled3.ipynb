```python
"""
End-to-End Text Classification Pipeline: Spam Detection
Build a complete ML pipeline with custom preprocessing, feature engineering,
Naive Bayes classification, and comprehensive performance evaluation.
"""

import string
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, classification_report
import pandas as pd

# ============================================================================
# DATASET
# ============================================================================

raw_dataset = [
    ("ham", "Go until jurong point, crazy.. Available only in bugis n great world la e buffet."),
    ("ham", "Ok lar... Joking wif u oni..."),
    ("spam", "Free entry in 2 a wkly comp to win FA Cup final tkts 21st May 2005. Text FA to 87121 to receive entry question(std txt rate)T&C's apply 08452810075over18's"),
    ("ham", "U dun say so early hor... U c already then say..."),
    ("ham", "Nah I don't think he goes to usf, he lives around here though"),
    ("spam", "FreeMsg Hey there darling it's been 3 week's now and no word back! I'd like some fun you up for it still? Txt chg to send £1.50 to rcv"),
    ("ham", "Even my brother is not like to speak with me. They treat me like aids patent."),
    ("ham", "As per your request 'Melle Melle (Oru Minnaminunginte Nurungu Vettam)' has been set as your callertune for all Callers. Press *9 to copy your friends Callertune"),
    ("spam", "WINNER!! As a valued network customer you have been selected to receivea £900 prize reward! To claim call 09061701461. Claim code KL341. Valid 12 hours only."),
    ("spam", "Had your mobile 11 months or more? U R entitled to Update to the latest colour mobiles with camera for Free! Call The Mobile Update Co FREE on 08002986030"),
    ("ham", "I'm gonna be home soon and i don't want to talk about this stuff any more tonight, k? I've cried enough today."),
    ("spam", "SIX chances to win CASH! From 100 to 2000 pounds txt> CSH11 and send to 87575. Cost 150p/day, 6days, 16+ TsandCs apply Reply HL 4 info"),
    ("spam", "URGENT! You have won a 1 week FREE membership in our £100,000 Prize Jackpot! Txt the word: CLAIM to No: 81010 T&C [www.dbuk.net] LCCLTD POBOX 4403LDNW1A7RW18"),
    ("ham", "I'VE BEEN SEARCHING FOR YOUR RIGHT WORDS FOR SO LONG IT FUNNY"),
    ("ham", "I HAVE A DATE ON SUNDAY WITH WILL!!"),
    ("spam", "XXXMobileMovieClub: To use your credit, click the WAP link in the next txt message or click here>> http://wap.xxxmobilemovieclub.com?n=QJKGIGHJJGCBL"),
    ("ham", "Oh k...i'm watching here:)"),
    ("ham", "Eh u remember how 2 spell his name... Yes i did. He v naughty make until i v wet."),
    ("ham", "Fine if that's the way u feel. That's the way its gota b"),
    ("spam", "England v Macedonia - dont miss the goals/team news. Txt ur national team to 87077 eg ENGLAND to 87077 Try:WALES, SCOTLAND 4txt/ú1.20 POBOXox36504W45WQ 16+")
]

# ============================================================================
# PHASE 1: TEXT PREPROCESSING & CLEANING (25% Weight)
# ============================================================================

def custom_processor(text):
    """
    Custom text preprocessing function that cleans raw strings.
    
    Operations performed:
    1. Convert all characters to lowercase
    2. Remove standard punctuation marks
    3. Normalize whitespace (remove extra, leading, trailing, and double spaces)
    
    Args:
        text (str): Raw input text string
    
    Returns:
        str: Cleaned and standardized text
    """
    # Step 1: Convert to lowercase
    text = text.lower()
    
    # Step 2: Remove all standard punctuation marks
    # Create a translation table mapping each punctuation character to None (deletion)
    translator = str.maketrans('', '', string.punctuation)
    text = text.translate(translator)
    
    # Step 3: Normalize whitespace
    # Split on whitespace (removes leading/trailing/multiple spaces) and rejoin with single space
    text = ' '.join(text.split())
    
    return text


# Test the custom_processor function
print("="*80)
print("PHASE 1: TEXT PREPROCESSING & CLEANING")
print("="*80)
print("\nTesting custom_processor() function:\n")

test_examples = [
    "Go until jurong point, crazy.. Available only in bugis n great world la e buffet.",
    "Free entry in 2 a wkly comp to win FA Cup final tkts 21st May 2005.",
    "I'm gonna be home soon and i don't want to talk about this stuff any more tonight, k?"
]

for example in test_examples:
    processed = custom_processor(example)
    print(f"Original: {example}")
    print(f"Cleaned:  {processed}")
    print()

# ============================================================================
# PHASE 2: FEATURE ENGINEERING & TOKENIZATION (25% Weight)
# ============================================================================

print("="*80)
print("PHASE 2: FEATURE ENGINEERING & TOKENIZATION")
print("="*80)

# Extract labels and texts from raw dataset
labels = [label for label, _ in raw_dataset]
texts = [text for _, text in raw_dataset]

# Initialize CountVectorizer with custom_processor as the preprocessing engine
vectorizer = CountVectorizer(preprocessor=custom_processor, lowercase=False)

# Transform cleaned text into numeric Bag-of-Words sparse matrix
X_sparse = vectorizer.fit_transform(texts)

print(f"\nCountVectorizer Configuration:")
print(f"  - Custom preprocessor: custom_processor()")
print(f"  - Vocabulary size: {len(vectorizer.get_feature_names_out())} unique words")
print(f"\nSparse Matrix Dimensions:")
print(f"  - Shape: {X_sparse.shape}")
print(f"  - Format: Compressed Sparse Row (CSR)")
print(f"  - Memory efficiency: {X_sparse.nnz} non-zero values out of {X_sparse.shape[0] * X_sparse.shape[1]} total elements")
print(f"  - Sparsity: {100 * (1 - X_sparse.nnz / (X_sparse.shape[0] * X_sparse.shape[1])):.2f}%")

print("\n" + "-"*80)
print("STRUCTURAL INTERPRETATION OF SPARSE MATRIX:")
print("-"*80)
print("""
Each Row (index 0 to 19):
  → Represents ONE text document from the dataset
  → Contains the word frequency counts for that specific message
  → Length: Equals vocabulary size (number of unique words across all documents)

Each Column (index 0 to vocabulary_size-1):
  → Represents ONE unique WORD from the vocabulary
  → Contains the frequency count of that word across all documents
  → Value at position [i, j] = count of word j in document i

Example:
  If Column 5 corresponds to the word "win" and Row 2 value is 3,
  this means document 2 contains the word "win" exactly 3 times.

Why Sparse?
  → Most documents don't contain most words
  → Storing only non-zero values saves memory
  → CSR format: Compressed Sparse Row (stores position + value pairs)
""")

# Display sample vocabulary and feature names
print("\nSample Vocabulary (first 20 words):")
feature_names = vectorizer.get_feature_names_out()
for idx, word in enumerate(feature_names[:20]):
    print(f"  Column {idx}: '{word}'")

# Display a sample document's features
print("\nSample: Document 2 (a spam message) - Word Counts:")
sample_doc_features = X_sparse[2].toarray()[0]
word_counts = [(feature_names[idx], sample_doc_features[idx]) 
               for idx in range(len(feature_names)) if sample_doc_features[idx] > 0]
word_counts.sort(key=lambda x: x[1], reverse=True)
for word, count in word_counts[:10]:
    print(f"  '{word}': {count}")

# ============================================================================
# PHASE 3: MODEL TRAINING & INFERENCE (25% Weight)
# ============================================================================

print("\n" + "="*80)
print("PHASE 3: MODEL TRAINING & INFERENCE")
print("="*80)

# Convert labels to numeric array (ham=0, spam=1)
y = np.array([1 if label == 'spam' else 0 for label in labels])

# Split dataset: 80% training, 20% testing with fixed random_state=42
X_train, X_test, y_train, y_test = train_test_split(
    X_sparse, y, test_size=0.2, random_state=42, stratify=y
)

print(f"\nTrain-Test Split (random_state=42):")
print(f"  - Training set size: {X_train.shape[0]} samples (80%)")
print(f"  - Test set size: {X_test.shape[0]} samples (20%)")
print(f"  - Training label distribution: {np.sum(y_train == 0)} ham, {np.sum(y_train == 1)} spam")
print(f"  - Test label distribution: {np.sum(y_test == 0)} ham, {np.sum(y_test == 1)} spam")

# Train MultinomialNB classifier
model = MultinomialNB()
model.fit(X_train, y_train)

print(f"\nMultinomialNB Model Trained Successfully")
print(f"  - Model parameters estimated from {X_train.shape[0]} training samples")
print(f"  - Feature space dimensionality: {X_train.shape[1]}")

# Inference on validation phrase
validation_phrase = "Win free cash prizes today!"
print(f"\n" + "-"*80)
print("INFERENCE ON VALIDATION PHRASE:")
print("-"*80)
print(f"Input: '{validation_phrase}'")

# Preprocess the validation phrase
processed_phrase = custom_processor(validation_phrase)
print(f"After preprocessing: '{processed_phrase}'")

# Vectorize using the fitted vectorizer
X_validation = vectorizer.transform([validation_phrase])

# Predict class
predicted_class = model.predict(X_validation)[0]
predicted_proba = model.predict_proba(X_validation)[0]

prediction_label = 'SPAM' if predicted_class == 1 else 'HAM'
print(f"\nPredicted Class: {prediction_label} (class index: {predicted_class})")
print(f"Class Probabilities:")
print(f"  - P(Ham): {predicted_proba[0]:.4f}")
print(f"  - P(Spam): {predicted_proba[1]:.4f}")

# ============================================================================
# PHASE 4: PERFORMANCE EVALUATION METRICS (25% Weight)
# ============================================================================

print("\n" + "="*80)
print("PHASE 4: PERFORMANCE EVALUATION METRICS")
print("="*80)

# Make predictions on test set
y_pred = model.predict(X_test)

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
print(f"\nConfusion Matrix:")
print(cm)
print(f"\nInterpretation:")
print(f"  - True Negatives (TN):  {cm[0, 0]} - Correctly identified Ham messages")
print(f"  - False Positives (FP): {cm[0, 1]} - Ham messages misclassified as Spam")
print(f"  - False Negatives (FN): {cm[1, 0]} - Spam messages misclassified as Ham")
print(f"  - True Positives (TP):  {cm[1, 1]} - Correctly identified Spam messages")

# Calculate metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

print(f"\nPerformance Metrics:")
print(f"  - Accuracy:  {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"  - Precision: {precision:.4f} ({precision*100:.2f}%)")
print(f"  - Recall:    {recall:.4f} ({recall*100:.2f}%)")
print(f"  - F1-Score:  {f1:.4f}")

print(f"\nDetailed Classification Report:")
print(classification_report(y_test, y_pred, target_names=['Ham', 'Spam'], zero_division=0))

# ============================================================================
# CRITICAL THINKING QUESTION: FALSE POSITIVE vs FALSE NEGATIVE
# ============================================================================

print("\n" + "="*80)
print("CRITICAL THINKING: FALSE POSITIVE vs FALSE NEGATIVE ANALYSIS")
print("="*80)

print(f"""
SCENARIO:
A spam filter incorrectly classifies a critical email from an employer as spam (FP)
versus allowing an advertisement email to slip into the inbox (FN).

ANALYSIS USING CALCULATED METRICS:

1. FALSE POSITIVE (Email from Employer → Spam Folder):
   ├─ Type: Legitimate message misclassified as Spam
   ├─ Current False Positive Rate: {cm[0,1]}/{cm[0,0]+cm[0,1]} from our test set
   ├─ Consequences:
   │  ├─ User misses critical job information/instructions
   │  ├─ Potential damage to professional relationship
   │  ├─ May miss time-sensitive opportunities or deadlines
   │  ├─ Could impact career advancement or compensation
   │  └─ High business impact, irreversible damage
   │
   └─ Severity: EXTREMELY HIGH (Lost opportunity, professional consequences)

2. FALSE NEGATIVE (Advertisement → Inbox):
   ├─ Type: Spam message misclassified as Legitimate
   ├─ Current False Negative Rate: {cm[1,0]}/{cm[1,0]+cm[1,1]} from our test set
   ├─ Consequences:
   │  ├─ User sees unwanted advertisement in inbox
   │  ├─ Minor annoyance, easily deleted
   │  ├─ Takes 2-3 seconds to identify and remove
   │  └─ Low business impact, easily recoverable
   │
   └─ Severity: LOW (Minor inconvenience, easily managed)

METRIC INSIGHTS:

Recall (True Positive Rate) = {recall:.4f}:
  → Shows our model catches {recall*100:.1f}% of actual spam messages
  → {(1-recall)*100:.1f}% of spam leaks through (False Negatives)
  → This is ACCEPTABLE because missing spam is better than blocking important mail

Precision (Positive Predictive Value) = {precision:.4f}:
  → When we mark something as spam, it's correct {precision*100:.1f}% of the time
  → {(1-precision)*100:.1f}% of flagged messages are actually legitimate (False Positives)
  → This is CRITICAL - we must minimize False Positives in email filtering

ANSWER:

⚠️  FALSE POSITIVE IS WORSE ⚠️

REASONING:
1. ASYMMETRIC COST: Blocking important email has orders of magnitude higher cost
   than showing ads. One missed employer email > 1000 ads in inbox.

2. TRUST & USABILITY: If the filter regularly blocks important emails, users
   disable it entirely (defeating the purpose). A few ads are tolerable.

3. OPTIMIZATION STRATEGY:
   - For email filtering, we should PRIORITIZE PRECISION (minimize False Positives)
   - Even if it means allowing more spam through (sacrificing Recall)
   - Set classification threshold > 0.5 to require higher confidence before blocking
   - Implement whitelist for known important senders (employers, banks, etc.)

4. METRIC TRADE-OFF:
   - Increasing Precision → Decreases Recall (more spam escapes)
   - Increasing Recall → Decreases Precision (more important emails blocked)
   - In email contexts, we choose the former trade-off

PRACTICAL CONFIGURATION:
Consider using a custom decision threshold (e.g., 0.75 instead of 0.5 for spam
classification) to reduce False Positives at the cost of more False Negatives.
""")

print("="*80)
print("END OF ANALYSIS")
print("="*80)
```
