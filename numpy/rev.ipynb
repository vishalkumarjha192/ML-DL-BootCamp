```python
import numpy as np

x=np.array([[[1,2],
             [3,4],
             [5,6]]])
print(x)

a=np.zeros((4,3))
print(a)

b=np.ones((3,4),dtype=int)
print(b)

```

```python
# broadcasting
a=np.array([1,2,3])
print(a+10)
print(a*2)

b=np.array([[12,58,78],
          [78,98,85]])
c=np.array([10,20,30])

print(b+c)

```
