```python
# Common imports for all problems
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes, make_regression
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X, y = make_regression(n_samples=100, n_features=1, noise=15, random_state=0)
 
model = LinearRegression().fit(X, y)
slope = model.coef_[0]
intercept = model.intercept_
 
print(f"Slope:     {slope:.3f}")
print(f"Intercept: {intercept:.3f}")
print(f"Interpretation: each 1-unit rise in x changes the prediction "
      f"by {slope:.2f}, starting from {intercept:.2f} when x = 0.")
```

```python
y_true = np.array([3.0, 5.0, 2.5, 7.0])
y_pred = np.array([2.8, 5.5, 2.0, 6.5])
 
mae  = np.mean(np.abs(y_true - y_pred))
mse  = np.mean((y_true - y_pred) ** 2)
rmse = np.sqrt(mse)
ss_res = np.sum((y_true - y_pred) ** 2)
ss_tot = np.sum((y_true - y_true.mean()) ** 2)
r2 = 1 - ss_res / ss_tot
 
print(f"Manual  -> MAE {mae:.4f} | MSE {mse:.4f} | RMSE {rmse:.4f} | R2 {r2:.4f}")
print(f"sklearn -> MAE {mean_absolute_error(y_true, y_pred):.4f} | "
      f"MSE {mean_squared_error(y_true, y_pred):.4f} | "
      f"RMSE {np.sqrt(mean_squared_error(y_true, y_pred)):.4f} | "
      f"R2 {r2_score(y_true, y_pred):.4f}")
```
