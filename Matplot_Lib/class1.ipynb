```python
import matplotlib.pyplot as plt
import numpy as np

x=np.linspace(0,2*np.pi,300)
y=np.sin(x)

plt.figure(figsize=(8,4))
plt.plot(x, y, color='red', linewidth=2, label='sin(x)')
plt.xlabel('x (radiands)')
plt.ylabel('Amplitude')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
 
rng = np.random.default_rng(42)
n = 200
 
# Generate 3 clusters
x1 = rng.normal(1, 0.5, n//3);  y1 = rng.normal(2, 0.5, n//3)
x2 = rng.normal(4, 0.5, n//3);  y2 = rng.normal(1, 0.5, n//3)
x3 = rng.normal(3, 0.5, n//3);  y3 = rng.normal(4, 0.5, n//3)
 
fig, ax = plt.subplots(figsize=(8, 7))

ax.scatter(x1, y1, c='Red', s=60, alpha=0.7, edgecolors='white',
           linewidths=0.5, label='Class 0', zorder=3)
ax.scatter(x2, y2, c='blue', s=60, alpha=0.7, edgecolors='white',
           linewidths=0.5, label='Class 1', zorder=3)
ax.scatter(x3, y3, c='green', s=60, alpha=0.7, edgecolors='white',
           linewidths=0.5, label='Class 2', zorder=3)
 
ax.set_title('Cluster Scatter Plot', fontsize=14, fontweight='bold')
ax.set_xlabel('Feature 1');  ax.set_ylabel('Feature 2')
ax.legend(framealpha=0.9)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
x = rng.uniform(0, 10, 150)
y = rng.uniform(0, 10, 150)
z = np.sin(x) * np.cos(y)       # 3rd variable drives colour
sizes = (rng.uniform(20, 200, 150))  # 4th variable drives size
 
fig, ax = plt.subplots(figsize=(8, 6))
sc = ax.scatter(x, y, c=z, s=sizes, cmap='plasma',
                alpha=0.8, edgecolors='none')
 
cbar = fig.colorbar(sc, ax=ax)
cbar.set_label('sin(x)·cos(y)', fontsize=11)
 
ax.set_title('Bubble Chart — 4 Variables', fontsize=15)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
 
categories = ['Model A', 'Model B', 'Model C', 'Model D', 'Model E']
accuracy   = [0.874, 0.921, 0.856, 0.943, 0.902]
errors     = [0.012, 0.008, 0.015, 0.006, 0.010]
 
fig, ax = plt.subplots(figsize=(9, 5))
 
colors = ['#9B59B6' if a < 0.9 else '#2ECC71' for a in accuracy]
bars = ax.bar(categories, accuracy, color=colors, width=0.6,
              yerr=errors, capsize=5, error_kw={'elinewidth': 1.5})
 
# Add value labels on top of bars
for bar, val in zip(bars, accuracy):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.005,
            f'{val:.3f}', ha='center', va='bottom', fontsize=10, fontweight='bold')
 
ax.set_title('Model Comparison — Test Accuracy', fontsize=14, fontweight='bold')
ax.set_ylabel('Accuracy', fontsize=12)
ax.set_ylim(0.8, 1.0)
ax.axhline(y=0.9, color='red', linestyle='--', alpha=0.5, label='Target: 0.90')
ax.legend()
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
rng = np.random.default_rng(0)
data1 = rng.normal(0, 1, 1000)
data2 = rng.normal(2, 1.5, 1000)
 
fig, ax = plt.subplots(figsize=(9, 5))
 
ax.hist(data1, bins=40, alpha=0.6, color='#3498DB',
        edgecolor='white', label='Group A')
ax.hist(data2, bins=40, alpha=0.6, color='#E74C3C',
        edgecolor='white', label='Group B')
 
# Add KDE-like vertical lines for means
ax.axvline(data1.mean(), color='#2980B9', linewidth=2, linestyle='--',
           label=f'Mean A: {data1.mean():.2f}')
ax.axvline(data2.mean(), color='#C0392B', linewidth=2, linestyle='--',
           label=f'Mean B: {data2.mean():.2f}')
 
ax.set_title('Overlapping Histograms', fontsize=14, fontweight='bold')
ax.set_xlabel('Value');  ax.set_ylabel('Frequency')
ax.legend()
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
```

```python
np.random.seed(42)
data = [np.random.normal(mu, std, 80)
        for mu, std in [(0,1),(1,1.5),(2,0.8),(3,2),(4,1.2)]]
 
fig, ax = plt.subplots(figsize=(9, 6))
 
bp = ax.boxplot(data, patch_artist=True, notch=True,
                medianprops={'color': 'white', 'linewidth': 2})
 
# Custom colours
colors = ['#9B59B6','#3498DB','#2ECC71','#E74C3C','#F39C12']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)
 
ax.set_xticklabels([f'G{i}' for i in range(1, 6)])
ax.set_title('Box Plot — Distribution Comparison', fontsize=14, fontweight='bold')
ax.set_ylabel('Value')
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
 
# Symmetric correlation matrix (6 features)
np.random.seed(42)
data = np.random.randn(100, 6)
corr = np.corrcoef(data.T)   # shape (6, 6)
 
features = ['Age', 'Income', 'Score', 'Tenure', 'Usage', 'Rating']
 
fig, ax = plt.subplots(figsize=(8, 7))
im = ax.imshow(corr, cmap='RdBu_r', vmin=-1, vmax=1, aspect='auto')
 
# Tick labels
ax.set_xticks(range(6));  ax.set_yticks(range(6))
ax.set_xticklabels(features, rotation=45, ha='right', fontsize=10)
ax.set_yticklabels(features, fontsize=10)
 
# Annotate each cell with the correlation value
for i in range(6):
    for j in range(6):
        text = ax.text(j, i, f'{corr[i,j]:.2f}',
                       ha='center', va='center', fontsize=9,
                       color='white' if abs(corr[i,j]) > 0.5 else 'black')
 
fig.colorbar(im, ax=ax, label='Pearson Correlation')
ax.set_title('Feature Correlation Matrix', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

```

```python
import matplotlib.pyplot as plt
import numpy as np
 
cm = np.array([[52, 3, 1],
               [2, 48, 4],
               [0, 2, 53]])
classes = ['Cat', 'Dog', 'Bird']
 
fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(cm, cmap='Blues')
 
ax.set_xticks(range(3));  ax.set_yticks(range(3))
ax.set_xticklabels(classes, fontsize=12)
ax.set_yticklabels(classes, fontsize=12)
ax.set_xlabel('Predicted Label', fontsize=12)
ax.set_ylabel('True Label', fontsize=12)
ax.set_title('Confusion Matrix', fontsize=14, fontweight='bold')
 
# Annotate with counts and percentages
total = cm.sum()
for i in range(3):
    for j in range(3):
        val = cm[i, j]
        ax.text(j, i, f'{val}\n({100*val/total:.1f}%)',
                ha='center', va='center', fontsize=11,
                color='white' if val > 40 else 'black',
                fontweight='bold')
fig.colorbar(im, ax=ax)
plt.tight_layout()
plt.show()


```
