```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import KNeighborsClassifier

# Dataset
data = pd.DataFrame({
    'Color': ['Red','Blue','Green','Red','Blue'],
    'Size': ['Small','Large','Medium','Large','Small'],
    'Shape': ['Round','Square','Round','Square','Round'],
    'Buy': ['Yes','No','Yes','No','Yes']
})

# Encode categorical features
encoders = {}

for col in data.columns:
    le = LabelEncoder()
    data[col] = le.fit_transform(data[col])
    encoders[col] = le

print(data)

# Features and target
X = data[['Color','Size','Shape']]
y = data['Buy']

# Train KNN
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X, y)

# New customer
new_data = pd.DataFrame({
    'Color':['Green'],
    'Size':['Small'],
    'Shape':['Round']
})

# Encode new data
for col in new_data.columns:
    new_data[col] = encoders[col].transform(new_data[col])

# Predict
prediction = knn.predict(new_data)

result = encoders['Buy'].inverse_transform(prediction)

print("Prediction:", result[0])
```

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
from sklearn.neighbors import KNeighborsClassifier

data = pd.DataFrame({
    'Color': ['Red','Blue','Green','Red','Blue'],
    'Size': ['Small','Large','Medium','Large','Small'],
    'Shape': ['Round','Square','Round','Square','Round'],
    'Class': ['Yes','No','Yes','No','Yes']
})
print(data)
X = data[['Color','Size','Shape']]
y = data['Class']

encoder = OrdinalEncoder()

X_encoded = encoder.fit_transform(X)
encoded_df = pd.DataFrame(
    X_encoded,
    columns=X.columns
)

print(encoded_df)
print(X_encoded)

knn = KNeighborsClassifier(
    n_neighbors=3,
    metric='hamming'
)

knn.fit(X_encoded, y)

new_point = pd.DataFrame({
    'Color':['Green'],
    'Size':['Small'],
    'Shape':['Round']
})

new_encoded = encoder.transform(new_point)

prediction = knn.predict(new_encoded)

print("Prediction:", prediction[0])
```
