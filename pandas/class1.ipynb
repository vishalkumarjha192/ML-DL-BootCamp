```python
import pandas as pd
import numpy as np

# 1D 
a=pd.Series([10,20,30])
print(a)

# index change
b=pd.Series([10,20,30],index=([1,2,3]))
print(b)

#np.nan -> not a mumber
case=pd.Series([10,20,np.nan])
print(case)
```

```python
a=[10,20,30,40,50]
print(a)
print(type(a))

b=pd.DataFrame(a,index=([1,2,4,5,7]))
print(b)
print(type(b))

```

```python
a=[["jay",89],["Raj",87],["shlok",98]]
print(a)
print(type(a))

b=pd.DataFrame(a)
print(b)
print(type(b))


b=pd.DataFrame(a,columns=["name","Marks"],index=[1,2,3])
print(b)
print(type(b))

print(b['name'])
print(b['Marks'])

r=b["name"]
print("Check : ",type(r))

c=b[["name"]]
print("Check angain : ",type(c))

print(b[["name","Marks"]])
```

```python
import numpy as np
import pandas as pd

a=np.array([[1,2],[3,4]])
print(a)
print(type(a))


b=pd.DataFrame(a,columns=["first","Second"],index=["first","Second"])
print(b)
print(type(b))
```

```python
a=pd.Series([178,22,3987,44,55,66],index=['a','b','c','d','e','f'])
print(a)
print(a+5)
print(a.mean())
print(a.var())
print(a.std())
print(a.max())
print(a.min())
print(a.value_counts())
print(a.count())
b=a[a<40]
print(b)
print(a.sort_values())
print(a.sort_values(ascending=True))
print(a[0:3])
print(a['a':'c'])

```
