```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Evan', 'Fiona'],
    'Department': ['HR', 'Engineering', 'Marketing', 'Engineering', 'HR', 'Sales'],
    'Age': [28, 35, 42, 29, 50, 31],
    'Salary': [60000, 95000, 80000, 105000, 65000, 72000],
    'Remote': [True, False, True, True, False, True],
    'Join_Date': ['2021-03-15', '2018-06-01', '2020-01-10', '2022-11-20', '2015-08-23', '2023-02-05']
}
df = pd.DataFrame(data)
df['Department']
df[['Name', 'Age', 'Salary']]
df[df['Remote'] == True]
df[df['Salary'] > 75000]
df[(df['Department'] == 'Engineering') & (df['Age'] < 32)]
df[df['Department'].isin(['HR', 'Sales'])]
df[(df['Department'] == 'HR') | (df['Department'] == 'Sales')]
df.iloc[0:3, 0:2]
df.loc[2:4, ['Name', 'Remote']]
df.loc[df['Salary'] < 70000, ['Name', 'Join_Date']]
```

```python
import pandas as pd

# Sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['NY', 'LA', 'Chicago', 'Houston']
}
df = pd.DataFrame(data, index=['a', 'b', 'c', 'd'])

# 1. Select a single row by label
print(df.loc['b'])

# 2. Select multiple rows
print(df.loc[['a', 'c']])

# 3. Select rows and specific columns
print(df.loc[['a', 'c'], ['Name', 'City']])

# 4. Select rows by label range (inclusive)
print(df.loc['a':'c'])

# 5. Conditional selection
print(df.loc[df['Age'] > 30])

# 6. Modify values using loc
df.loc['b', 'Age'] = 31
print(df)

# 7. Assign multiple values
df.loc['d', ['Age', 'City']] = [45, 'Dallas']
print(df)
```

```python
import numpy as np
import pandas as pd

data = {
    'Department': ['IT', 'IT', 'HR', 'HR', np.nan, 'Sales', 'Sales'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'Diana', 'Evan', 'Fiona', 'George'],
    'Salary': [100000, np.nan, 60000, 65000, 85000, np.nan, 70000]
}

df = pd.DataFrame(data)
#print(df)
print(df.groupby('Department')['Salary'].sum())
print(df.groupby('Department', dropna=False)['Salary'].sum())
stats = df.groupby("Department")["Salary"].agg(['sum', 'mean', 'count'])#for multiple aggregations at once, use .agg():
print(stats)
df['Dept_Mean'] =df.groupby('Department')['Salary'].transform('mean')
#print(df)
df['Clean_Salary'] = df['Salary'].fillna(df['Dept_Mean'])
print(df)
```

```python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Gender_Code': ['F', 'M', 'F']
})

# 2. Define the mapping dictionary
gender_map = {
    'M': 'Male',
    'F': 'Female'
}

# 3. Apply the map
df['Gender'] = df['Gender_Code'].map(gender_map)
print(df)
```

```python
import pandas as pd

# Parse dates from string
df = pd.DataFrame({
    'date': ['2024-01-15', '2024-03-22', '2024-07-04', '2024-11-28'],
    'value': [120, 135, 98, 145]
})
df['date'] = pd.to_datetime(df['date'])


# Extract components
df['year']    = df['date'].dt.year
df['month']   = df['date'].dt.month
df['day']     = df['date'].dt.day
df['weekday'] = df['date'].dt.day_name()  # 'Monday', 'Tuesday'...
df['quarter'] = df['date'].dt.quarter

print(df.head())
```

```python
# Vertical stack (add rows) — same columns
q1 = pd.DataFrame({'month': ['Jan','Feb'], 'revenue': [100, 120]})
q2 = pd.DataFrame({'month': ['Apr','May'], 'revenue': [150, 130]})

full_year = pd.concat([q1, q2], ignore_index=True)
print(full_year)

# Horizontal stack (add columns) — same index
features = pd.DataFrame({'A': [1,2,3], 'B': [4,5,6]})
labels   = pd.DataFrame({'y': [0,1,0]})

dataset = pd.concat([features, labels], axis=1)
print(dataset)

```
