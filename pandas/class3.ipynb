```python
import pandas as pd
import numpy as np

data = {
    'Department': ['IT', 'IT', 'HR', 'HR', np.nan, 'Sales', 'Sales'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'Diana', 'Evan', 'Fiona', 'George'],
    'Salary': [100000, np.nan, 60000, 65000, 85000, np.nan, 70000]
}

df = pd.DataFrame(data)

# Groupby sum
print(df.groupby('Department')['Salary'].sum())

# Groupby with NaN included
print(df.groupby('Department', dropna=False)['Salary'].sum())

# Calculate department mean
df['Dept_Mean'] = df.groupby('Department')['Salary'].transform('mean')

# Fill missing salary with department mean
df['Clean Salary'] = df['Salary'].fillna(df['Dept_Mean'])

# Final output
print(df)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame([[8,16],[4,9],[49,21]], columns=['A','B'])
print("Original DataFrame:\n", df)

# Apply lambda to multiply by 2
df_double = df.apply(lambda x: x * 2)
print("\nAfter multiplying by 2:\n", df_double)

# Column-wise sum
df1 = df.apply(np.sum, axis=0)
print("\nColumn-wise sum:\n", df1)

# Row-wise sum
df2 = df.apply(np.sum, axis=1)
print("\nRow-wise sum:\n", df2)
```

```python
import pandas as pd

# Load CSV file
csv = pd.read_csv("Data (1).csv")

# Display dataset
print("Full Dataset:\n", csv)

# 1. Select first row
print("\nFirst Row:\n", csv.iloc[0])

#2. Select specific value (row 0, column 2 -> Salary)
print("\nSpecific Value:", csv.iloc[0, 2])

#3. Select all rows, Age column
print("\nAge Column:\n", csv.iloc[:, 1])

#4. Select Country & Age columns
print("\nCountry & Age:\n", csv.iloc[:, 0:2])

#5. First 5 rows
print("\nFirst 5 Rows:\n", csv.iloc[0:5])

#6. Rows + Columns together
print("\nSubset (0:3 rows, 0:3 columns):\n", csv.iloc[0:3, 0:3])

#7. Features (X) and Target (y)
X = csv.iloc[:, :-1].values   # all columns except last
y = csv.iloc[:, -1].values    # last column

print("\nFeatures (X):\n", X)
print("\nTarget (y):\n", y)

#8. Specific rows & columns
print("\nSelected Rows & Columns:\n", csv.iloc[[1, 3, 5], [0, 2]])
```
