```python
import pandas as pd

# 1. Create a sample dataset of student exam scores
data = {'Student': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Henry'],
        'Score': [55, 62, 67, 80, 91, 93, 97, 99]}
df = pd.DataFrame(data)

print("--- Original Data ---")
print(df)


# 2. Equal-Width Binning (using pd.cut)
# This divides the score range (from min to max) into equal intervals.
# We will create 3 bins: Low, Medium, and High scores.
df['Equal_Width_Bins'] = pd.cut(df['Score'], bins=3, labels=['Low', 'Medium', 'High'])


# 3. Equal-Frequency Binning (using pd.qcut)
# This divides the data so that an equal number of students fall into each bin.
# We will create 2 bins (quantiles/halves): 'Bottom Half' and 'Top Half'.
df['Equal_Freq_Bins'] = pd.qcut(df['Score'], q=2, labels=['Bottom Half', 'Top Half'])


print("\n--- After Binning ---")
print(df)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# 1. Create sample dataset
data = {'Pet Name': ['Buddy', 'Whiskers', 'Bella', 'Max'],
        'Animal Type': ['Dog', 'Cat', 'Bird', 'Dog']}
df = pd.DataFrame(data)

# 2. Initialize the OneHotEncoder
# 'sparse_output=False' ensures it returns a regular NumPy array instead of a sparse matrix
encoder = OneHotEncoder(sparse_output=False)

# 3. Fit and transform the categorical column
# We reshape or select it as a 2D array [['Animal Type']] because sklearn expects 2D inputs
encoded_array = encoder.fit_transform(df[['Animal Type']])
print(encoded_array)

# 4. Get the new feature column names automatically
column_names = encoder.get_feature_names_out(['Animal Type'])
print(column_names)

# 5. Convert back into a readable DataFrame
encoded_df = pd.DataFrame(encoded_array, columns=column_names, dtype=int)
print(encoded_df)

# Combine with the original 'Pet Name' column
final_df = pd.concat([df['Pet Name'], encoded_df], axis=1)

print(final_df)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. Create sample dataset
data = {'Product': ['Shirt A', 'Shirt B', 'Shirt C', 'Shirt D'],
        'Size': ['Large', 'Small', 'Medium', 'Large']}
df = pd.DataFrame(data)

# 2. Initialize the LabelEncoder
le = LabelEncoder()

# 3. Fit and transform the column
# LabelEncoder expects a 1D array, so we pass df['Size'] with single brackets
df['Size_Encoded'] = le.fit_transform(df['Size'])

print(df)
```
