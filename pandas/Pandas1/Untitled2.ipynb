```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('train.csv')


# 1. Handle Missing Values

df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df['Fare'] = df['Fare'].fillna(df['Fare'].mean())


# 2. Subset Selection

df = df[['PassengerId', 'Name', 'Age', 'Fare', 'Embarked', 'Pclass']]


# 3. Binning (Age)
def age_group(age):
    if age < 13:
        return 'Child'
    elif age < 20:
        return 'Teenager'
    elif age < 60:
        return 'Adult'
    else:
        return 'Senior'

df['Age_Group'] = df['Age'].apply(age_group)


# 4. Log Transformation (Fare)

df['Fare_log'] = np.log1p(df['Fare'])  # log(1 + x)


# 5. One-Hot Encoding (Embarked)

df = pd.get_dummies(df, columns=['Embarked'], drop_first=True)


# Final Output

print(df.head())
```

```python
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer
from reportlab.lib.styles import getSampleStyleSheet

# Create document
doc = SimpleDocTemplate("ML_Metrics_Notes.pdf")
styles = getSampleStyleSheet()

content = []

# Title
content.append(Paragraph("<b>Machine Learning Evaluation Metrics Notes</b>", styles["Title"]))
content.append(Spacer(1, 12))

# Accuracy
content.append(Paragraph("<b>1. Accuracy</b>", styles["Heading2"]))
content.append(Paragraph("Ratio of correctly predicted observations to total observations.", styles["Normal"]))
content.append(Paragraph("<b>Formula:</b> (TP + TN) / (TP + TN + FP + FN)", styles["Normal"]))
content.append(Spacer(1, 10))

# Precision
content.append(Paragraph("<b>2. Precision</b>", styles["Heading2"]))
content.append(Paragraph("Measures how many predicted positives are actually correct.", styles["Normal"]))
content.append(Paragraph("<b>Formula:</b> TP / (TP + FP)", styles["Normal"]))
content.append(Spacer(1, 10))

# Recall
content.append(Paragraph("<b>3. Recall</b>", styles["Heading2"]))
content.append(Paragraph("Measures how many actual positives are correctly identified.", styles["Normal"]))
content.append(Paragraph("<b>Formula:</b> TP / (TP + FN)", styles["Normal"]))
content.append(Spacer(1, 10))

# F1 Score
content.append(Paragraph("<b>4. F1 Score</b>", styles["Heading2"]))
content.append(Paragraph("Harmonic mean of Precision and Recall.", styles["Normal"]))
content.append(Paragraph("<b>Formula:</b> 2 × (Precision × Recall) / (Precision + Recall)", styles["Normal"]))
content.append(Spacer(1, 10))

# Confusion Matrix
content.append(Paragraph("<b>5. Confusion Matrix</b>", styles["Heading2"]))
content.append(Paragraph("Table used to evaluate classification models.", styles["Normal"]))
content.append(Paragraph("TP: True Positive, TN: True Negative, FP: False Positive, FN: False Negative", styles["Normal"]))
content.append(Spacer(1, 10))

# ROC Curve
content.append(Paragraph("<b>6. ROC Curve</b>", styles["Heading2"]))
content.append(Paragraph("Graph of True Positive Rate vs False Positive Rate.", styles["Normal"]))
content.append(Paragraph("TPR = TP / (TP + FN)", styles["Normal"]))
content.append(Paragraph("FPR = FP / (FP + TN)", styles["Normal"]))
content.append(Spacer(1, 10))

# AUC Score
content.append(Paragraph("<b>7. AUC Score</b>", styles["Heading2"]))
content.append(Paragraph("Area under ROC curve.", styles["Normal"]))
content.append(Paragraph("Range: 0.5 (random) to 1.0 (perfect)", styles["Normal"]))
content.append(Spacer(1, 10))

# Build PDF
doc.build(content)

print("PDF Created Successfully!")
```
