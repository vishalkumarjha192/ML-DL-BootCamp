```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
# !. Load and scale Data
df=pd.read_csv("Iris.csv")

# print(df[['SepalLengthCm','SepalWidthCm']].head())

numeric_cols = df.select_dtypes(include=['int64', 'float64'])

scaler=MinMaxScaler()
scaled_values = scaler.fit_transform(numeric_cols)

df1=pd.DataFrame(scaled_values, columns=numeric_cols.columns)

# View the Scaled Columns

print("...scaled data preview......")
print(df1[['SepalLengthCm','SepalWidthCm']].head())

```

```python
import pandas as pd
from sklearn.preprocessing import MaxAbsScaler

# Load data
df = pd.read_csv("Iris.csv")

# Drop categorical column
df.drop('Species', axis=1, inplace=True)

print("--- DataFrame Before MaxAbsScaler (First 5 rows) ---")
print(df.head())

# Apply MaxAbsScaler
scaler = MaxAbsScaler()
scaled_data = scaler.fit_transform(df)

# Convert to DataFrame
df1 = pd.DataFrame(scaled_data, columns=df.columns)

print("--- After MaxAbsScaler ---")
print(df1.head())
```
