```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

n=500

# Generate Age
age=np.random.randint(20,60,n)
# print(age)

# Correlate income
income=age*2000+np.random.normal(0,10000,n)
# print(income)

# Categories
category = np.random.choice(['A', 'B', 'C'], n)
# print(category)

df = pd.DataFrame({
    'age': age,
    'income': income,
    'category': category
})
# print(df)

group_mean=df.groupby('category')['income'].mean()
# print(group_mean)


# bar chart

plt.figure(figsize=(7,5))
group_mean.plot(kind='bar')
plt.title("mean income by category")
plt.ylabel("mean income")
plt.grid(axis='y')
plt.show()


# scatter plot

plt.figure(figsize(8,6))

for cat in df['category'].unique():
    subset=df[df['category'==cat]]
    plt.scatter(subset)
```
