```python
# Question1

import numpy as np
import time

arr= np.random.randint(0, 100, size=(1000, 5))
print("Reading array")
print(arr)

print("\nArray Attributes:")
print("Shape:", arr.shape)
print("Number of Dimensions:", arr.ndim)
print("Total Elements:", arr.size)
print("Data Type:", arr.dtype)

bro = np.array([1, 2, 3, 4, 5])
bro_read = arr + bro
print("\nFirst 5 Rows After Broadcasting:")
print(bro_read[:5])


```

```python
import numpy as np
import time
sensor_readings = np.random.randint(0, 100, size=(1000, 5))

print("Sensor Readings Array:")
print(sensor_readings[:5])

print("\nArray Attributes:")
print("Shape:", sensor_readings.shape)
print("Number of Dimensions:", sensor_readings.ndim)
print("Total Elements:", sensor_readings.size)
print("Data Type:", sensor_readings.dtype)

calibration = np.array([1, 2, 3, 4, 5])

calibrated_readings = sensor_readings + calibration

print("\nFirst 5 Rows After Broadcasting:")
print(calibrated_readings[:5])


avg_readings = np.mean(sensor_readings, axis=0)

print("\nAverage Reading of Each Sensor:")
print(avg_readings)

N = 1_000_000
numpy_array = np.arange(N)

start = time.time()
numpy_result = numpy_array * 2
numpy_time = time.time() - start

# Python List
python_list = list(range(N))

start = time.time()
loop_result = []

for x in python_list:
    loop_result.append(x * 2)

loop_time = time.time() - start

print("\nPerformance Comparison")
print(f"NumPy Vectorized Time : {numpy_time:.6f} seconds")
print(f"Python Loop Time      : {loop_time:.6f} seconds")
print(f"Speedup               : {loop_time / numpy_time:.2f}x faster")
```
