```python
"""
Leaf Classification using Transfer Learning (EfficientNetB0)
Dataset: csafrit2/plant-leaves-for-image-classification (Kaggle)

Two-phase training:
1. Feature extraction — freeze pretrained base, train only the new head
2. Fine-tuning — unfreeze top layers of the base, train with a low LR
"""

import os
import kagglehub
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# ----------------------------------------------------------------------
# 1. Download dataset
# ----------------------------------------------------------------------
path = kagglehub.dataset_download("csafrit2/plant-leaves-for-image-classification")
print("Path to dataset files:", path)

def find_split_dirs(root):.
    found = {}
    for dirpath, dirnames, _ in os.walk(root):
        for d in dirnames:
            key = d.lower()
            if key in ("train", "valid", "validation", "test"):
                found[key] = os.path.join(dirpath, d)
    return found

split_dirs = find_split_dirs(path)
print("Detected splits:", split_dirs)

train_dir = split_dirs.get("train")
val_dir = split_dirs.get("valid") or split_dirs.get("validation")
test_dir = split_dirs.get("test")

# ----------------------------------------------------------------------
# 2. Build datasets
# ----------------------------------------------------------------------
IMG_SIZE = (224, 224)   # EfficientNetB0's native input size
BATCH_SIZE = 32
SEED = 42

train_ds = tf.keras.utils.image_dataset_from_directory(
    train_dir, image_size=IMG_SIZE, batch_size=BATCH_SIZE, seed=SEED
)
val_ds = tf.keras.utils.image_dataset_from_directory(
    val_dir, image_size=IMG_SIZE, batch_size=BATCH_SIZE, seed=SEED
) if val_dir else None
test_ds = tf.keras.utils.image_dataset_from_directory(
    test_dir, image_size=IMG_SIZE, batch_size=BATCH_SIZE, seed=SEED
) if test_dir else None

class_names = train_ds.class_names
num_classes = len(class_names)
print(f"Found {num_classes} classes:", class_names)

if val_ds is None:
    val_batches = int(0.2 * tf.data.experimental.cardinality(train_ds).numpy())
    val_ds = train_ds.take(val_batches)
    train_ds = train_ds.skip(val_batches)

AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.cache().shuffle(1000).prefetch(AUTOTUNE)
val_ds = val_ds.cache().prefetch(AUTOTUNE)
if test_ds:
    test_ds = test_ds.cache().prefetch(AUTOTUNE)

data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05),
], name="data_augmentation")
# Dataset is very small (~5 images/class in test, similarly few in
# train/val), so augmentation was scaled way down from the original
# (rotation 0.15->0.05, zoom/contrast removed). Heavy augmentation on
# a tiny dataset can distort images enough that the model struggles
# to learn stable features — less is more here.

# ----------------------------------------------------------------------
# 3. Build model: EfficientNetB0 backbone + custom head
# ----------------------------------------------------------------------
# Note: EfficientNet's preprocess_input is a no-op — normalization
# (rescaling to the range the pretrained weights expect) is built
# into the model's first layers already, so raw 0-255 pixel values
# go in directly.
base_model = tf.keras.applications.EfficientNetB0(
    input_shape=IMG_SIZE + (3,),
    include_top=False,
    weights="imagenet",
)
base_model.trainable = False  # Phase 1: freeze the pretrained backbone

inputs = tf.keras.Input(shape=IMG_SIZE + (3,))
x = data_augmentation(inputs)
x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
x = layers.Dense(128, activation="relu")(x)
x = layers.Dropout(0.2)(x)
outputs = layers.Dense(num_classes, activation="softmax")(x)

model = tf.keras.Model(inputs, outputs, name="leaf_efficientnetb0")
model.summary()

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

callbacks = [
    tf.keras.callbacks.EarlyStopping(patience=5, restore_best_weights=True, monitor="val_loss"),
    tf.keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, min_lr=1e-7),
    tf.keras.callbacks.ModelCheckpoint("best_leaf_efficientnetb0.keras", save_best_only=True, monitor="val_loss"),
]

# ----------------------------------------------------------------------
# 4. Phase 1 — train the new head only (backbone frozen)
# ----------------------------------------------------------------------
EPOCHS_HEAD = 15
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS_HEAD,
    callbacks=callbacks,
)

# ----------------------------------------------------------------------
# 5. Phase 2 — fine-tune: unfreeze the top layers of the backbone
# ----------------------------------------------------------------------
base_model.trainable = True

# Freeze everything except the last ~30 layers so early, generic
# feature detectors (edges, textures) stay intact and only the
# more task-specific layers adapt.
FINE_TUNE_AT = len(base_model.layers) - 30
for layer in base_model.layers[:FINE_TUNE_AT]:
    layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),  # much lower LR
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

EPOCHS_FINE_TUNE = 15
total_epochs = EPOCHS_HEAD + EPOCHS_FINE_TUNE

history_fine = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=total_epochs,
    initial_epoch=history.epoch[-1] + 1,
    callbacks=callbacks,
)

# ----------------------------------------------------------------------
# 6. Combine + plot training curves across both phases
# ----------------------------------------------------------------------
def plot_history(h1, h2, switch_epoch):
    acc = h1.history["accuracy"] + h2.history["accuracy"]
    val_acc = h1.history["val_accuracy"] + h2.history["val_accuracy"]
    loss = h1.history["loss"] + h2.history["loss"]
    val_loss = h1.history["val_loss"] + h2.history["val_loss"]

    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    plt.plot(acc, label="Train Accuracy")
    plt.plot(val_acc, label="Val Accuracy")
    plt.axvline(switch_epoch, linestyle="--", color="gray", label="Fine-tuning starts")
    plt.legend()
    plt.title("Accuracy")

    plt.subplot(1, 2, 2)
    plt.plot(loss, label="Train Loss")
    plt.plot(val_loss, label="Val Loss")
    plt.axvline(switch_epoch, linestyle="--", color="gray", label="Fine-tuning starts")
    plt.legend()
    plt.title("Loss")
    plt.tight_layout()
    plt.savefig("training_curves_transfer.png", dpi=150)
    plt.show()

plot_history(history, history_fine, switch_epoch=len(history.epoch))

# ----------------------------------------------------------------------
# 7. Evaluate on test set
# ----------------------------------------------------------------------
if test_ds:
    test_loss, test_acc = model.evaluate(test_ds)
    print(f"Test accuracy: {test_acc:.4f} | Test loss: {test_loss:.4f}")

    y_true, y_pred = [], []
    for images, labels in test_ds:
        preds = model.predict(images, verbose=0)
        y_true.extend(labels.numpy())
        y_pred.extend(np.argmax(preds, axis=1))

    print(classification_report(y_true, y_pred, target_names=class_names))

    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=False, cmap="Blues", xticklabels=class_names, yticklabels=class_names)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title("Confusion Matrix (EfficientNetB0 Transfer Learning)")
    plt.tight_layout()
    plt.savefig("confusion_matrix_transfer.png", dpi=150)
    plt.show()

# ----------------------------------------------------------------------
# 8. Save final model
# ----------------------------------------------------------------------
model.save("leaf_efficientnetb0_final.keras")
print("Model saved as leaf_efficientnetb0_final.keras")
```
